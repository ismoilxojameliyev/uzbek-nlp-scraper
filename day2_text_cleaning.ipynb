```python
# Sample raw text simulating messy scraped content
raw_scraped_text = "   O'zbekiston   Respublikasi   Prezidenti   matbuot   xizmati... \n\n   "

# 1. Strip trailing and leading whitespace
trimmed = raw_scraped_text.strip()

# 2. Split by any whitespace and rejoin to normalize irregular spaces
normalized_words = trimmed.split()
clean_text = " ".join(normalized_words)

print("Original:", repr(raw_scraped_text))
print("Cleaned: ", repr(clean_text))
```

```python
# Mapping inconsistent quote marks to a single standardized apostrophe
char_replacement_map = {
    "‘": "'",
    "’": "'",
    "`": "'",
    "ʻ": "'",
    "ʼ": "'",
}

sample_sentence = "Oʻzbekiston va Qoraqalpog`iston o‘rtasidagi do‘stlik aloqalari."

normalized_sentence = sample_sentence
for bad_char, clean_char in char_replacement_map.items():
    normalized_sentence = normalized_sentence.replace(bad_char, clean_char)

print("Before:", sample_sentence)
print("After: ", normalized_sentence)
```

```python
import requests
from bs4 import BeautifulSoup
import time

# 1. Create a Python list containing multiple URLs
urls_to_scrape = [
    "https://daryo.uz/",
    "https://kun.uz/",
    "https://gazeta.uz/oz/"
]

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/111.0.0.0 Safari/537.36"
}

# 2. Start the loop to visit each URL one by one
for current_url in urls_to_scrape:
    print(f"Connecting to: {current_url}")

    # Send the request
    response = requests.get(current_url, headers=headers)

    # Check if successful
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        paragraphs = soup.find_all('p')

        # Extract and print just the first 3 paragraphs to keep output clean
        for p in paragraphs[:3]:
            clean_text = p.get_text(strip=True)
            if clean_text:
                print(clean_text)
    else:
        print(f"Failed to connect. Status: {response.status_code}")

    print("--------------------------------------------------")

    # 3. Pause the script for 2 seconds before the next loop
    time.sleep(2)
```

```python
import requests
from bs4 import BeautifulSoup
import time
import json  # Import the JSON module

urls_to_scrape = [
    "https://daryo.uz/",
    "https://kun.uz/",
    "https://gazeta.uz/oz/"
]

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/111.0.0.0 Safari/537.36"
}

# 1. Create an empty list to hold all the scraped data
dataset = []

for current_url in urls_to_scrape:
    print(f"Scraping: {current_url}")
    response = requests.get(current_url, headers=headers)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        paragraphs = soup.find_all('p')

        # Combine the paragraphs into one single block of text
        article_text = ""
        for p in paragraphs[:3]:
            clean_text = p.get_text(strip=True)
            if clean_text:
                article_text += clean_text + " "

        # 2. Add the URL and text into our list as a dictionary
        dataset.append({
            "source": current_url,
            "text": article_text.strip()
        })

    time.sleep(2)

# 3. Create a new file and write the data into it
with open("uzbek_nlp_data.json", "w", encoding="utf-8") as json_file:
    json.dump(dataset, json_file, ensure_ascii=False, indent=4)

print("Scraping complete. Data saved to uzbek_nlp_data.json!")
```
